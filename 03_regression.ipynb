# Análisis de regresión

## Objetivo

Construir un modelo de regresión lineal múltiple que explique el PIB per cápita a partir de variables socioeconómicas.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

df = pd.read_csv(
    r"C:\Users\Alfi\Desktop\analista de datos\estadística\proyecto estadística\proyecto-socioeconómico\scripts\socioeconomic_dataset_clean.csv"
)
```

```python
df['log_gdp'] = np.log(df['gdp_per_capita'])
```

```python
X = df[['life_expectancy', 'internet_usage', 'unemployment']]
X = sm.add_constant(X)

y = df['log_gdp']
```

```python
model = sm.OLS(y, X).fit()
print(model.summary())
```

## Interpretación del modelo

El modelo de regresión lineal múltiple presenta un R² de 0,719, lo que indica que las variables independientes incluidas explican aproximadamente el 71,9 % de la variabilidad observada en el logaritmo del PIB per cápita dentro de la muestra.

La esperanza de vida presenta un coeficiente positivo (0,1394) y estadísticamente significativo (p < 0,05), lo que indica una asociación positiva con el logaritmo del PIB per cápita, manteniendo constantes las demás variables del modelo.

El uso de internet también presenta un coeficiente positivo (0,0113) y estadísticamente significativo (p < 0,05), lo que indica una asociación positiva con la variable dependiente, controlando por esperanza de vida y desempleo.

Por otro lado, el desempleo presenta un coeficiente cercano a cero (0,0005) y no estadísticamente significativo (p = 0,829). Por tanto, no se encuentra evidencia estadística suficiente de una asociación lineal entre el desempleo y el logaritmo del PIB per cápita una vez consideradas las demás variables del modelo.

## Limitaciones del modelo

El modelo permite identificar asociaciones estadísticas entre las variables socioeconómicas y el logaritmo del PIB per cápita, pero no establece relaciones causales. Entre sus limitaciones se encuentran la posible omisión de variables relevantes y la existencia de multicolinealidad entre las variables independientes.

Además, el conjunto de datos contiene observaciones de distintos países a lo largo de varios años, por lo que pueden existir dependencias entre observaciones de un mismo país. El modelo OLS convencional no considera explícitamente esta estructura de datos, lo que puede afectar la validez de los errores estándar y de las pruebas de significación estadística.

Por último, el R² obtenido representa el ajuste dentro de la muestra analizada y no garantiza el mismo desempeño predictivo con datos nuevos.
