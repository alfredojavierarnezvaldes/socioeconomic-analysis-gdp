# Hypothesis Testing

## Objetivo

Evaluar la relación estadística entre variables socioeconómicas y el PIB per cápita mediante el coeficiente de correlación de Pearson.

```python
import pandas as pd
from scipy.stats import pearsonr

df = pd.read_csv(
    r"C:\Users\Alfi\Desktop\analista de datos\estadística\proyecto estadística\proyecto-socioeconómico\scripts\socioeconomic_dataset_clean.csv"
)
```

```python
import numpy as np

df['log_gdp'] = np.log(df['gdp_per_capita'])
```

## Hipótesis: Esperanza de vida vs PIB per cápita

H0: No existe relación  
H1: Existe relación

```python
corr, p_value = pearsonr(df['life_expectancy'], df['log_gdp'])
print(corr, p_value)
```

Se encontró una correlación positiva fuerte (r ≈ 0.82) y estadísticamente significativa (p < 0.05), lo que indica una relación robusta entre ambas variables.

## Hipótesis: Uso de internet vs PIB per cápita

```python
corr, p_value = pearsonr(df['internet_usage'], df['log_gdp'])
print(corr, p_value)
```

Correlación positiva moderada (r ≈ 0.66), significativa.

## Hipótesis: Desempleo vs PIB per cápita

```python
corr, p_value = pearsonr(df['unemployment'], df['log_gdp'])
print(corr, p_value)
```

Correlación negativa muy débil (r ≈ -0.10), aunque estadísticamente significativa, con bajo poder explicativo.

## Conclusiones

Los resultados muestran que la esperanza de vida presenta la correlación positiva más fuerte con el logaritmo del PIB per cápita (r = 0,822), seguida por el uso de internet (r = 0,661). En contraste, el desempleo presenta una correlación negativa muy débil (r = -0,106), aunque su valor p es inferior a 0,05.

Estas asociaciones no implican causalidad. Además, dado que el conjunto de datos incluye observaciones de los mismos países en diferentes años, los valores p deben interpretarse con cautela, ya que las pruebas convencionales de Pearson suponen independencia entre observaciones.
