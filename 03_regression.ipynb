# Análisis de regresión

## Objetivo

Construir un modelo de regresión lineal múltiple que explique el PIB per cápita a partir de variables socioeconómicas.

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

df = pd.read_csv('C:/Users/Alfi/Desktop/analista de datos/estadística/proyecto estadística/proyecto-socioeconómico/data/clean/final_dataset.csv')
```

```python
df['log_gdp'] = np.log(df['gdp_per_capita'])
```

```python
X = df[['life_expectancy', 'internet_usage', 'unemployment']]
X = sm.add_constant(X)

y = df['log_gdp']
```

```python
model = sm.OLS(y, X).fit()
print(model.summary())
```

## Interpretación del modelo

El modelo presenta un R² de aproximadamente 0.72, lo que indica que explica una proporción considerable de la variabilidad del PIB per cápita.

La esperanza de vida muestra el mayor coeficiente positivo, lo que sugiere que es la variable con mayor impacto en el modelo.

El uso de internet también presenta un efecto positivo y estadísticamente significativo, aunque de menor magnitud.

Por otro lado, el desempleo no resulta estadísticamente significativo, lo que indica que no aporta poder explicativo relevante cuando se consideran las demás variables.

## Limitaciones del modelo

El modelo es de naturaleza correlacional y no implica causalidad. Además, pueden existir variables omitidas y multicolinealidad entre variables independientes.
